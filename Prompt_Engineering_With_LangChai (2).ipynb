# 7.2 Conversation Buffer Window Memory

# Keeps the last n turns of the conversation. Drops the oldest turn when the buffer is full.

```python
from langchain_core.messages import trim_messages

selected_messages = trim_messages(
    history.messages,
    token_counter=len,  # <-- len will simply count the number of messages rather than tokens
    max_tokens=3,  # <-- allow up to 2 messages.
    strategy="last",
    # Most chat models expect that chat history starts with either:
    # (1) a HumanMessage or
    # (2) a SystemMessage followed by a HumanMessage
    # start_on="human" makes sure we produce a valid chat history
    start_on="human",
    # Usually, we want to keep the SystemMessage
    # if it's present in the original history.
    # The SystemMessage has special instructions for the model.
    include_system=True,
    allow_partial=False,
)

for msg in selected_messages:
    msg.pretty_print()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, AIMessage
from langgraph.graph import StateGraph, END

gemini_api_key = gemini_api_key

model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite",
    google_api_key=gemini_api_key,
    temperature=0.7,
)

# --- STEP 1: Define the state ---
from typing import List, TypedDict

class State(TypedDict):
    messages: List
    summary: str

# --- STEP 2: Define your summarization reducer ---
def summarize_conversation(state: State):
    summary = state.get("summary", "")
    if summary:
        summary_message = (
            f"This is a summary of the conversation so far: {summary}\n\n"
            "Extend the summary by considering the new messages above."
        )
    else:
        summary_message = "Create a summary of the conversation above."

    # Add the summarization instruction to the messages
    messages = state["messages"] + [HumanMessage(content=summary_message)]
    response = model.invoke(messages)

    # Keep the last 2 messages and store new summary
    return {
        "summary": response.content,
        "messages": state["messages"][-2:]  # keep short memory
    }

# --- STEP 3: Define the main conversation node ---
def chat_node(state: State):
    response = model.invoke(state["messages"])
    return {"messages": state["messages"] + [response]}

# --- STEP 4: Build the graph ---
graph = StateGraph(State)

graph.add_node("chat", chat_node)
graph.add_node("summarize", summarize_conversation)
graph.add_edge("chat", "summarize")
graph.add_edge("summarize", END)

graph.set_entry_point("chat")

app = graph.compile()

# --- STEP 5: Run the conversation ---
state = {"messages": [HumanMessage(content="List 3 places to visit in Paris.")], "summary": ""}

state = app.invoke(state)
print("Bot:", state["messages"][-1].content)
print("Summary:", state["summary"])

# Continue conversation
state["messages"].append(HumanMessage(content="Which one of them has a lake view?"))
state = app.invoke(state)

print("\nBot:", state["messages"][-1].content)
print("Updated summary:", state["summary"])
```

# 7.6 Load Saved Messages into history

```python
from langchain_core.messages import messages_from_dict
from langchain_community.chat_message_histories import ChatMessageHistory
import json

# Load From Json
with open("Conversation-memory.json" , "r") as f:
  dicts = json.load(f)
```

```python
# Convert to Message objects
messages = messages_from_dict(dicts)

# Create ChatMessageHistory and load them
history = ChatMessageHistory()
for msg in messages:
    history.add_message(msg)

print("Loaded history:")
for m in history.messages:
    print(f"{m.type}: {m.content}")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables.history import RunnableWithMessageHistory

chat = ChatGoogleGenerativeAI(
      model = "gemini-2.5-flash-lite",
      temperature = 0.7 ,
      google_api_key = gemini_api_key
)
store = {"session_1": history}  # reuse loaded history

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

chain = RunnableWithMessageHistory(chat, get_session_history)

response = chain.invoke(
    "as an egyptian, which place from the 3 i will like most?",
    config={"configurable": {"session_id": "session_1"}},
)

print(response.content)
```

## 12.2 MapReduce chain (LCEL 2025)





```python
map_prompt = PromptTemplate(
    input_variables =['document'],
    template = """
    You are given a document chunk. summarize the core meaning consisely.

    Document:
    {document}

    Summary:
    """
)

reduce_prompt = PromptTemplate(
    input_variables=["summaries","question"],
    template="""
    You are given multiple summaries. use them to answer the question
    if the answer is not contained in the summaries, say "No Answer is Available"
    Summaries:
   {summaries}

   Question:
   {question}

   Answer:
    """
)
```

```python
def map_reduce_chain(docs, question):
  summaries = []
  for doc in docs:
    summaries.append(chat.invoke(map_prompt.format(document=doc.page_content)).text.strip())
  return chat.invoke(reduce_prompt.format(summaries = summaries , question = question)).text.strip()
```

```python
question = "what did Tesla invent?"

similar_docs = vector_db.similarity_search(question, k=3)
```

```python
answer = map_reduce_chain(similar_docs, question)
print(answer)
```

## 12.3 Refine chain (LCEL 2025)

---



```python
initial_prompt = PromptTemplate(
    input_variables = ["document", "question"],
    template="""
    Answer the following question using the provided document,

    Document:
    {document}

    Question:
    {question}

    Answer:
    """
)

refine_prompt = PromptTemplate(
    input_variables = ["existing_answer", "document", "question"],
    template="""
    We have an exsiting answer:
    {existing_answer}

    Here is new information from another document:
    {document}

    Question:
    {question}

    Refine the existing answer to include this ew information
    If the new document does not add anything, keep the answer unchanged

    Refined Answer:
    """
)
```

```python
def refine_chain(docs, question):
  # initial step (given best match doc)
  answer = chat.invoke(initial_prompt.format(document=docs[0].page_content, question=question)).text.strip()
  # refine step (given rest of docs)
  for doc in docs[1:]:
    answer =  chat.invoke(refine_prompt.format(existing_answer=answer,
                                                        document=doc.page_content,
                                                        question=question)).text.strip()
  return answer
```

```python
question = "what did Tesla invent?"

similar_docs = vector_db.similarity_search(question, k=2)
```

```python
answer = refine_chain(similar_docs, question)
print(answer)
```
